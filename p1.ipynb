```python
n=int(input())#Two sum problem in list
arr=[]
found=False
for i in range(n):
    num=int(input())
    arr.append(num)
print(arr)
target=int(input())
print(target)
for i in range(n):
    for j in range(i+1,n):
        if arr[i]+arr[j]==target:
            print(i,j)
            found=True
if found==False:
    print("not")
```

```python
row =int(input())#Sum the row and column in list
column=int(input())
mat=[]
total=0
for i in range(row):
    arr=[]
    for j in range(column):
        n=int(input())
        arr.append(n)
    mat.append(arr)
print(mat)
for i in range(row):
    for j in range(column):
        if i==j:
            total=total+mat[i][j]
print(total)
```
