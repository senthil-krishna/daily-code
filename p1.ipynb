```python
num1= int(input())
num2=int(input())

sum_a = 0
sum_b = 0

for i in range(1, num1):
    if num1 % i == 0:
        sum_a = sum_a + i

for i in range(1, num2):
    if num2 % i == 0:
        sum_b = sum_b + i
if sum_a == num2 and sum_b == num1:
    print(num1, "and", num2 , "are Amicable Numbers")
else:
    print(num1, "and", num2, "are not Amicable Numbers")
```

```python
n=int(input()) #Frame pattern
for row in range(1,n+1):
    for column in range(1,n+1):
        if row==1 or row==n or column==1 or column==n:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```
